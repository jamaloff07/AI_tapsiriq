Temperatur məlumatları əsasında əsas göstəriciləri hesablamalısınız. İcra qaydaları:

Kod xanasında temperatur siyahısı yaradın:  [2, 5, 4, 6, 7, 3, 1] .
Python funksiyalarından istifadə edərək aşağıdakıları hesablayın:
1. orta qiymət;
2. maksimum və minimum qiymətlər;
3. maksimum və minimum qiymətlər arasındakı fərq.

Nəticələri cədvəl şəklində təqdim edin (Markdown).

```python

temperatur = [2, 5, 4, 6, 7, 3, 1]

# Orta qiymətin hesablanması
orta = sum(temperatur) / len(temperatur)

# Maksimum və minimum qiymətlərin tapılması
maksimum = max(temperatur)
minimum = min(temperatur)

# Maksimum və minimum arasındakı fərq
ferq = maksimum - minimum

# Nəticələrin göstərilməsi
print("Temp siyahi:", temperatur)
print("Orta qiymet:", orta)
print("Max temperatur:", maksimum)
print("Min temperatur:", minimum)
print("Max ve min ferq:", ferq)
```
