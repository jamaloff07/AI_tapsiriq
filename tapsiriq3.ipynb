# Tapşırıq 3 — Səslərin siyahısını emal etməli və lideri avtomatik müəyyən etməlisiniz.

Bu tapşırıqda meyvələrin səslər siyahısında neçə dəfə təkrarlandığını hesablayacağıq.

**Tapşırığın məqsədləri:**
 - Hər meyvənin neçə dəfə təkrarlandığını hesablamaq üçün Python-un daxili vasitələrindən (məsələn, count(), dict və ya collections.Counter) istifadə edin.
 - Ən çox rast gəlinən meyvəni — lideri — müəyyən edən kod yazın.
 - Nəticələri “meyvə → say” formatında cədvəldə təqdim edin.
 - İlk 3 nəticəni ayrıca göstərin.



```python

from collections import Counter

votes = [
    "apple", "banana", "apple", "orange",
    "banana", "apple", "apple", "orange"
]

counts = Counter(votes)

print("MEYVE → SAY")
print("----------------")

for meyve, say in counts.most_common():
    print(f"{meyve} → {say}")

lider, lider_sayi = counts.most_common(1)[0]

print("\nLider:", lider)
print("Liderin ses sayi:", lider_sayi)

ilk_3 = counts.most_common(3)

print("\nİLK 3 NETİCE")
for meyve, say in ilk_3:
    print(f"{meyve} → {say}")
```
