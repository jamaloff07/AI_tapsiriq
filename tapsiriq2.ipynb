
# Tapşırıq 2 — Gündəlik xərclər siyahısı üzrə göstəriciləri hesablamalısınız.


**Hesablanacaq göstəricilər:**
 - xərclərin ümumi məbləği;
 - orta qiymət;
 - ən yüksək və ən aşağı xərc;
 - xərcin 1000-dən çox olduğu günlərin sayı.



```python

expenses = [1200, 1500, 800, 900, 1000, 450, 900]

umumi_xerc = sum(expenses)
orta_xerc = sum(expenses) / len(expenses)

en_yuksek = max(expenses)
en_asagi = min(expenses)

say = 0

for xerc in expenses:
    if xerc > 1000:
        say += 1

print("gunler uzre xerc:", expenses)
print("umumi xerc:", umumi_xerc)
print("orta xerc:", orta_xerc)
print("max xerc:", en_yuksek)
print("min xerc:", en_asagi)
print("1000 manatdan cox xerc olan gunlerin sayi:", say)
```
