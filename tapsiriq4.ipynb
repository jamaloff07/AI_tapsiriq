# Tapşırıq 4 — Fənlər üzrə nəticələri müqayisə edin və əsas göstəriciləri hesablayın.


**Hesablanacaq göstəricilər:**
- Hər fənn üzrə orta bal.
- Hər fənn üzrə minimum bal.
- Hər fənn üzrə maksimum bal.
- Orta nəticəsi ən yüksək olan fənn.

```python

math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]

scores = {
    "Math": math,
    "Physics": physics,
    "English": english
}

averages = {}

for subject, marks in scores.items():
    averages[subject] = sum(marks) / len(marks)

    print(f"\nFənn: {subject}")
    print(f"Orta bal: {averages[subject]:.2f}")
    print(f"Min bal: {min(marks)}")
    print(f"Max bal: {max(marks)}")

best_subject = max(averages, key=averages.get)

print("\nen yuksek orta bal:", best_subject)
```
